## Step 1: Load the Data

Import the libraries. `pandas` holds the data, `seaborn` and `matplotlib` make the graphs, and `sklearn` (scikit-learn) does the machine learning.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

Load the California housing dataset straight from the web. `usecols` keeps only the three columns we need. `.head()` shows the first 5 rows.

```python
url = "https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv"
home_data = pd.read_csv(url, usecols=["longitude", "latitude", "median_house_value"])
home_data.head()
```

## Step 2: Look at the Data

Plot every neighborhood by its location. `hue` colors each dot by its house value.

```python
sns.scatterplot(data=home_data, x="longitude", y="latitude", hue="median_house_value")
```

## Step 3: Scale

`StandardScaler` puts `latitude` and `longitude` on the same range. `fit_transform` learns the range and scales the data in one step. `X` is the scaled data K-Means will use.

```python
scaler = StandardScaler()
X = scaler.fit_transform(home_data[["latitude", "longitude"]])
```

**Why does scaling matter for K-Means?**

*Double-click this cell and replace this line with your answer.*

## Step 4: Run K-Means with K = 3

Create a K-Means model that looks for 3 clusters, then train it on `X`. `random_state=0` makes your results match your classmates'.

```python
kmeans = KMeans(n_clusters=3, random_state=0, n_init="auto")
kmeans.fit(X)
```

`kmeans.labels_` holds the cluster number for every row. Save it as a new column.

```python
home_data["cluster"] = kmeans.labels_
```

Make the same plot as Step 2, but color by cluster.

```python
sns.scatterplot(data=home_data, x="longitude", y="latitude", hue="cluster", palette="tab10")
```

## Step 5: Compare Prices

A box plot shows the spread of house values in each cluster. The line in the middle of each box is the median.

```python
sns.boxplot(data=home_data, x="cluster", y="median_house_value")
```

**Do the clusters have different house prices? How can you tell?**

## Step 6: Choose K

Train a model for every K from 1 to 10. Save each model's **inertia** (how spread out its clusters are) in a list.

```python
K = range(1, 11)
inertias = []
for k in K:
    model = KMeans(n_clusters=k, random_state=0, n_init="auto").fit(X)
    inertias.append(model.inertia_)
```

Plot inertia vs. K. Look for the **elbow**, where the line stops dropping quickly.

```python
sns.lineplot(x=list(K), y=inertias, marker="o")
plt.xlabel("K (number of clusters)")
plt.ylabel("Inertia")
```

Check the silhouette score for a few K values near your elbow. Higher is better. Change the list to the K values you want to test. `sample_size=5000` speeds it up.

```python
for k in [3, 4, 5, 6]:
    model = KMeans(n_clusters=k, random_state=0, n_init="auto").fit(X)
    score = silhouette_score(X, model.labels_, sample_size=5000, random_state=0)
    print(f"K = {k}: silhouette score = {score:.3f}")
```

## Step 7: Rerun with Your K

Set `best_k` to the K you chose, then train a new model with it and save the new labels.

```python
best_k = 5  # Change this to the K you chose

kmeans = KMeans(n_clusters=best_k, random_state=0, n_init="auto")
kmeans.fit(X)
home_data["cluster"] = kmeans.labels_
```

Redo your cluster map.

```python
sns.scatterplot(data=home_data, x="longitude", y="latitude", hue="cluster", palette="tab10")
```

Redo your box plot.

```python
sns.boxplot(data=home_data, x="cluster", y="median_house_value")
```
